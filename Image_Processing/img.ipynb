# Rotate

```python
image_r = image.rotate(30)
```

```python
image_r.show()
```

```python
image_r = image.rotate(60)
image_r.save('image_r.png')
```

```python
image_r = image.rotate(60)
image_r.show()
```

```python
image_r = image.rotate(-60,expand=True, fillcolor=(0,255,255))
image_r.show()
```

```python
image_r = image.rotate(-60,expand=True, fillcolor=(100,100,100))
image_r.show()
```

```python
cropped = image.crop((200,300,300,500))
cropped.show()
```

# Flipping the Image

```python
image_h=image.transpose(Image.Transpose.FLIP_LEFT_RIGHT)
image_h.show()
```
