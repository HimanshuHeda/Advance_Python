# Image Enhancer - Brightness, Sharpness, color and Contrast

```python
from PIL import ImageEnhance
```

```python
image_col=ImageEnhance.Color(image)
colored_image=image_col.enhance(0.5)
colored_image.show()
```

```python
image_con=ImageEnhance.Contrast(image)
contrasted_image=image_con.enhance(1.20)
contrasted_image.show()
```

```python
image_sharp=ImageEnhance.Sharpness(image)
sharped_image=image_sharp.enhance(500)
sharped_image.show()
```

```python
image_bright=ImageEnhance.Brightness(image)
brightened_image=image_bright.enhance(1)
brightened_image.show()
```

# Image Filters 

```python
from PIL import ImageFilter
```

```python
blur_image=image.filter(ImageFilter.BLUR)
blur_image.show()
```

```python
contour_image=image.filter(ImageFilter.CONTOUR)
contour_image.show()
```

```python
detail_image=image.filter(ImageFilter.DETAIL)
detail_image.show()
```

```python
edge_enhance_image=image.filter(ImageFilter.EDGE_ENHANCE)
edge_enhance_image.show()
```

```python
edge_enchancemore_image=image.filter(ImageFilter.EDGE_ENHANCE_MORE)
edge_enchancemore_image.show()
```

```python
emboss_image=image.filter(ImageFilter.EMBOSS)
emboss_image.show()
```

```python
find_edges_image=image.filter(ImageFilter.FIND_EDGES)
find_edges_image.show()
```

```python
sharpen_image=image.filter(ImageFilter.SHARPEN)
sharpen_image.show()
```

```python
smooth_image=image.filter(ImageFilter.SMOOTH)
smooth_image.show()
```

```python
smooth_more_image=image.filter(ImageFilter.SMOOTH_MORE)
smooth_more_image.show()
```
