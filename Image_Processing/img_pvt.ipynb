## Image Attributes

```python
print(image.size)  # (width, height)
print(image.mode)  # e.g., "RGB", "RGBA", "L" (grayscale)
print(image.format)  # e.g., "JPEG", "PNG"
```

```python
from PIL import Image, ImageDraw, ImageFont

# Open the image
image = Image.open("./shuttlecock.jpg")

# Create a drawing object
draw = ImageDraw.Draw(image)

# Load a font (ensure the font file exists in your system or directory)
# You can use a default font if you don't have a specific font file
try:
    font = ImageFont.truetype("arial.ttf", size=40)  # Use a TrueType font
except IOError:
    font = ImageFont.load_default()  # Fallback to default font

# Add text to the image
draw.text((50, 50), "Hello, World!", fill="white", font=font)

# Show the image
image.show()
```

```python
image.save("output_image.jpg", format="JPEG")
```

```python
width, height = image.size
crop = image.crop((0, 0, width // 2, height // 2))  # Valid crop
crop.show()
```

#  Advanced Topics

## Image Blending

```python
image1 = Image.open("./code_01.jpg")
image2 = Image.open("./code_02.jpg")
```

```python
image2 = image2.resize(image1.size)
```

```python
blended_image = Image.blend(image1, image2, alpha=0.5)
blended_image.show()
```

## Image Masking

```python
mask = Image.open("./code_01.jpg").convert("L")
masked_image = Image.composite(image1, image2, mask)
masked_image.show()
```

## Image Enhancements 

```python
from PIL import ImageEnhance
enhancer = ImageEnhance.Brightness(image)
enhanced_image = enhancer.enhance(1.5)  # Increase brightness by 1.5x
enhanced_image.show()
```
