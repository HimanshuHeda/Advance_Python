# 2. Make the Image Squared image

```python
min_dimension = min(image.size)
squared_image = image.crop((0, 0, min_dimension, min_dimension))
squared_image.show()
squared_image.save("squared-kitten.jpg")
```

# 3. Consider the original image and crop it to get only the face of the kitten.

```python
crop_box = (200, 50, 400, 250)  
cropped_image = image.crop(crop_box)
cropped_image.show()
cropped_image.save("cropped-kitten.jpg")
```

# 4. Apply filter

### Import the Library for ImageFiltering

```python
from PIL import ImageFilter
```

```python
filtered_image = image.filter(ImageFilter.CONTOUR)
filtered_image.show()
filtered_image.save("contour-kitten.jpg")
```

```python
blur_image=image.filter(ImageFilter.BLUR)
blur_image.show()
blur_image.save("blur-kitten.jpg")
```

```python
detail_image=image.filter(ImageFilter.DETAIL)
detail_image.show()
detail_image.save("detail-kitten.jpg")
```

```python
edge_enhance_image=image.filter(ImageFilter.EDGE_ENHANCE)
edge_enhance_image.show()
edge_enhance_image.save("edge-enhance-kitten.jpg")
```

```python
edge_enchancemore_image=image.filter(ImageFilter.EDGE_ENHANCE_MORE)
edge_enchancemore_image.show()
edge_enchancemore_image.save("edge-enhance-more-kitten.jpg")
```

```python
sharpen_image=image.filter(ImageFilter.SHARPEN)
sharpen_image.show()
sharpen_image.save("sharpen-kitten.jpg")
```

```python
smooth_image=image.filter(ImageFilter.SMOOTH)
smooth_image.show()
smooth_image.save("smooth-kitten.jpg")
```

```python
smooth_image=image.filter(ImageFilter.SMOOTH_MORE)
smooth_image.show()
smooth_image.save("smooth-more-kitten.jpg")
```

# 5 Flip and rotate the image, customize

```python
image_h=image.transpose(Image.Transpose.FLIP_LEFT_RIGHT)
image_h.show()
image_h.save("flip-kitten.jpg")
```

```python
image_h=image.transpose(Image.Transpose.FLIP_TOP_BOTTOM)
image_h.show() 
image_h.save("flip-kitten1.jpg")
```

# 6 Apply enchancement features

### Importing Libraries 

```python
from PIL import ImageEnhance
```

```python
image_col=ImageEnhance.Color(image)
colored_image=image_col.enhance(1)
colored_image.show()
colored_image.save("colored-kitten.jpg")
```

```python
image_con=ImageEnhance.Contrast(image)
contrasted_image=image_con.enhance(1.20)
contrasted_image.show()
contrasted_image.save("contrasted-kitten.jpg")
```

```python
image_sharp=ImageEnhance.Sharpness(image)
sharped_image=image_sharp.enhance(500)
sharped_image.show()
sharped_image.save("sharped-kitten.jpg")
```

```python
image_bright=ImageEnhance.Brightness(image)
brightened_image=image_bright.enhance(1)
brightened_image.show()
brightened_image.save("brightened-kitten.jpg")
```
